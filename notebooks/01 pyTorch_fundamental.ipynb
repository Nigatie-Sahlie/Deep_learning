### **Manupilating tesnor(Operation)**

```python
vect = torch.tensor([1,4,6])
print(vect + 10, vect * 10, vect / 10)
print(vect)
```

```python
vector1 = torch.tensor([1,4,8])
vector2 = torch.tensor([10, 31, 10])
print(torch.matmul(vector1, vector2))    #final answer are given one element(number)
print(vector1.shape, vector2.shape)
```

**More about shape of Tensor**

```python
#random tensor
TENSOR_A = torch.rand(size=(3, 8))
TENSOR_B = torch.rand(size=(3, 8))
TENSOR_A.shape, TENSOR_A, TENSOR_B.shape, TENSOR_B
```

```python
#Transpose of tesnors
TENSOR_B.shape, TENSOR_B, TENSOR_B.T.shape, TENSOR_B.T
```

```python
import torch.nn as nn 

input_data = torch.rand(size=(3, 4))
print(f"input_data shape: {input_data.shape}")
layer = nn.Linear(in_features=4, out_features=6)
print(f"layer shape: {layer.weight.shape}")
output = layer(input_data)
print(f"output shape: {output.shape}")
output

#the inpout inner and layer inner must have the same shape
```

**Reshaping**

```python
print(TENSOR_B, TENSOR_B.shape, TENSOR_B.ndim)
reshaped_TENSOR_B = TENSOR_B.reshape(6, 4)
#or
reshaped_TENSOR_B1 = torch.reshape(TENSOR_B, (6, 4))
print(reshaped_TENSOR_B, reshaped_TENSOR_B.shape, reshaped_TENSOR_B.ndim)
print(reshaped_TENSOR_B1, reshaped_TENSOR_B1.shape, reshaped_TENSOR_B1.ndim)
```

```python
B = torch.arange(1., 13.)
print(B, B.shape, B.ndim)
view_B = B.view(1, 12)
reshap_B = B.reshape(4, 3)
print(view_B, view_B.shape, view_B.ndim)
view_B[:, 0] = 5
reshap_B[:, 0] = 10
view_B, B, reshap_B, B
```

```python
#Then stacked 
stacked = torch.stack([B,B,B])
B, B.shape, B.ndim, stacked, stacked.shape, stacked.ndim
```

**Selecting data from Tensor(Indexing)**

```python
#Switch between numpy and pyTorch
print(B, type(B))
array_B = B.numpy()
print(array_B, type(array_B))

import numpy as np
np_array = np.arange(1., 13.)
print(np_array, type(np_array))

to_Tensor = torch.from_numpy(np_array)
print(to_Tensor, type(to_Tensor))  #The data type is defualt float64 as numpy array and we can change float32 as defualt to tensor
```
